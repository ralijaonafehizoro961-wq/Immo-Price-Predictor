```python
import joblib
import numpy as np
import pandas as pd
```

```python
saved = joblib.load("../models/house_model.joblib")
pipeline = saved["pipeline"]
features = saved["features"]
```

```python
house = {
    "sqft_living": 1180,
    "sqft_lot": 5650,
    "bedrooms": 3,
    "bathrooms": 1.0,
    "floors": 1.0,
    "waterfront": 0,
    "view": 0,
    "condition": 3,
    "grade": 7,
    "age": 60,
    "renovated": 0,
    "zipcode": "98178",
}
```

```python
row = pd.DataFrame([house])[features]
price = np.exp(pipeline.predict(row)[0])
print("Estimated price:", round(price), "dollars")
print("Real price: 221900 dollars")
```
