```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/kc_house_data.csv")
```

```python
print("Before:", len(df))
```

```python
clean = df[(df["bedrooms"] > 0)& (df["bathrooms"] > 0)& (df["bedrooms"] < 10)].copy()
```

```python
print("After:", len(clean))
print("Removed:", len(df) - len(clean))
```

```python
clean.to_csv("../data/processed/kc_house_clean.csv", index=False)
print("Saved.")
```
