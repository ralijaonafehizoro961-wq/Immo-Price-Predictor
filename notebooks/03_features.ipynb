```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("../data/processed/kc_house_clean.csv")
print("Rows:", len(df))
```

```python
df["age"] = 2015 - df["yr_built"]
df["renovated"] = (df["yr_renovated"] > 0).astype(int)
df["log_price"] = np.log(df["price"])
```

```python
FEATURES = [
    "sqft_living",
    "sqft_lot",
    "bedrooms",
    "bathrooms",
    "floors",
    "waterfront",
    "view",
    "condition",
    "grade",
    "age",
    "renovated",
    "zipcode",
]
TARGET = "log_price"
```

```python
data = df[FEATURES + ["price", TARGET]].copy()
```

```python
print("Columns:", data.columns.tolist())
print("Empty cells:", data.isna().sum().sum())
```

```python
links = data.drop(columns=["zipcode", TARGET]).corr()["price"]
print(links.sort_values(ascending=False))
```

```python
data.to_csv("../data/processed/kc_house_features.csv", index=False)
print("Saved.")
```
