```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
FILE_PATH = '../data/raw/kc_house_data.csv'
df = pd.read_csv(FILE_PATH)
```

```python
print("Rows and columns:", df.shape)
```

```python
df.head()
```

```python
df.info()
```

```python
print("Empty cells:", df.isna().sum().sum())
```

```python
print(df["price"].describe())
```

```python
print("Cheapest:", df["price"].nsmallest(5).tolist())
print("Most expensive:", df["price"].nlargest(5).tolist())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(df["price"], bins=50)
axes[0].set_title("Price")

axes[1].hist(np.log(df["price"]), bins=50)
axes[1].set_title("Log of price")
plt.show()
```
